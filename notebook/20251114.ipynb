## 前処理の方針

確認した結果に基づいて、以下の前処理を実装：
1. **欠損値処理**
   - **50%以上欠損の列は削除**
   - **数値変数の欠損値はMLで予測して補完**（LightGBMを使用）
   - **カテゴリ変数の欠損値は補完しない**（そのまま残す）
2. **カテゴリ変数のエンコーディング**
   - カーディナリティが低い場合はOne-Hot Encoding
   - カーディナリティが高い場合はTarget Encodingや頻度エンコーディング
3. **数値変数の変換**
   - 必要に応じて対数変換やスケーリング
4. **特徴量選択**
   - 相関の高い特徴量を優先
   - 重要度の低い特徴量を削除


## 前処理パイプラインの構築


```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.impute import SimpleImputer
from sklearn.metrics import mean_squared_error, mean_absolute_error
import warnings
warnings.filterwarnings('ignore')

# lightgbmのインポート
try:
    import lightgbm as lgb
    print("LightGBM imported successfully")
except ImportError:
    print("LightGBM not found. Installing...")
    import subprocess
    subprocess.check_call(["pip", "install", "lightgbm"])
    import lightgbm as lgb

# 50%以上欠損の列を削除
missing_threshold = 50.0
cols_to_drop = missing_df[missing_df['欠損率(%)'] >= missing_threshold].index.tolist()
print(f"削除する列（{missing_threshold}%以上欠損）: {len(cols_to_drop)}列")
print(f"削除する列名（最初の10個）: {cols_to_drop[:10]}")  # 最初の10個のみ表示

# 目的変数とID列を分離
y = df['money_room'].copy()
X = df.drop(['money_room'] + cols_to_drop, axis=1, errors='ignore')

print(f"\n前処理後の特徴量数: {X.shape[1]}")
print(f"サンプル数: {X.shape[0]}")
```

```python
# MLで欠損値補完を行うクラスを定義
class MLImputer:
    """機械学習を使って欠損値を補完するクラス"""
    
    def __init__(self, model_type='lgb', random_state=42):
        self.model_type = model_type
        self.random_state = random_state
        self.models = {}  # 各列ごとのモデルを保存
        self.imputed_columns = []  # 補完した列のリスト
        
    def fit_transform(self, X_train, X_test=None):
        """
        trainデータで学習し、trainとtestの両方の欠損値を補完
        """
        X_train_imputed = X_train.copy()
        X_test_imputed = X_test.copy() if X_test is not None else None
        
        # 欠損値がある列を特定
        missing_cols = X_train.columns[X_train.isnull().any()].tolist()
        
        print(f"欠損値補完が必要な列数: {len(missing_cols)}")
        
        # 各欠損列について順番に補完
        for i, col in enumerate(missing_cols):
            # カテゴリ変数（object型）は補完しない
            col_dtype = X_train_imputed[col].dtype
            if col_dtype == 'object' or col_dtype.name == 'object':
                print(f"[{i+1}/{len(missing_cols)}] 列 '{col}' はカテゴリ変数（{col_dtype}）のため補完をスキップします")
                continue
            
            # 数値型でない場合はスキップ
            if col_dtype not in ['int64', 'float64', 'int32', 'float32']:
                print(f"[{i+1}/{len(missing_cols)}] 列 '{col}' は数値型ではない（{col_dtype}）ため補完をスキップします")
                continue
            
            print(f"[{i+1}/{len(missing_cols)}] 列 '{col}' の欠損値を補完中...")
            
            # 欠損値のインデックスを取得
            train_missing_idx = X_train_imputed[col].isnull()
            if X_test_imputed is not None and col in X_test_imputed.columns:
                test_missing_idx = X_test_imputed[col].isnull()
            else:
                test_missing_idx = pd.Series([False] * len(X_test_imputed)) if X_test_imputed is not None else pd.Series(dtype=bool)
            
            # 欠損値がない場合はスキップ
            if not train_missing_idx.any() and (X_test_imputed is None or not test_missing_idx.any()):
                continue
            
            # 特徴量として使用する列（目的列以外で、欠損率が低い数値列のみ）
            feature_cols = []
            for c in X_train_imputed.columns:
                if c == col:
                    continue
                # 数値列のみを使用
                if X_train_imputed[c].dtype not in ['int64', 'float64', 'int32', 'float32']:
                    continue
                # その列の欠損率が50%未満の列のみ使用
                missing_rate = X_train_imputed[c].isnull().sum() / len(X_train_imputed)
                if missing_rate < 0.5:
                    feature_cols.append(c)
            
            if len(feature_cols) == 0:
                print(f"  警告: 列 '{col}' の補完に使用できる特徴量がありません。中央値で補完します。")
                # フォールバック: 中央値で補完（数値変数のみ）
                fill_value = X_train_imputed[col].median()
                if pd.notna(fill_value):
                    X_train_imputed.loc[train_missing_idx, col] = fill_value
                    if X_test_imputed is not None and col in X_test_imputed.columns:
                        X_test_imputed.loc[test_missing_idx, col] = fill_value
                continue
            
            # 学習データの準備
            train_data = X_train_imputed[feature_cols + [col]].copy()
            
            # 欠損値がある行を一時的に削除（特徴量の欠損も考慮）
            train_clean = train_data.dropna(subset=feature_cols + [col])
            
            if len(train_clean) < 100:  # データが少なすぎる場合はスキップ
                print(f"  警告: 列 '{col}' の学習データが少なすぎます（{len(train_clean)}行）。中央値で補完します。")
                fill_value = X_train_imputed[col].median()
                if pd.notna(fill_value):
                    X_train_imputed.loc[train_missing_idx, col] = fill_value
                    if X_test_imputed is not None and col in X_test_imputed.columns:
                        X_test_imputed.loc[test_missing_idx, col] = fill_value
                continue
            
            # 特徴量の前処理（数値列のみなので、欠損値を0で埋めるだけ）
            train_processed = train_clean[feature_cols].fillna(0).copy()
            
            # 目的変数
            y_train = train_clean[col].copy()
            
            # モデルの学習
            try:
                model = self._train_model(train_processed, y_train, col)
                self.models[col] = model
            except Exception as e:
                print(f"  エラー: モデルの学習に失敗しました: {e}")
                print(f"  中央値で補完します。")
                fill_value = X_train_imputed[col].median()
                if pd.notna(fill_value):
                    X_train_imputed.loc[train_missing_idx, col] = fill_value
                    if X_test_imputed is not None and col in X_test_imputed.columns:
                        X_test_imputed.loc[test_missing_idx, col] = fill_value
                continue
            
            # trainデータの欠損値を予測
            if train_missing_idx.any():
                try:
                    train_missing_data = X_train_imputed.loc[train_missing_idx, feature_cols].copy()
                    train_missing_processed = train_missing_data.fillna(0)
                    predictions = self._predict(model, train_missing_processed, col)
                    X_train_imputed.loc[train_missing_idx, col] = predictions
                except Exception as e:
                    print(f"  警告: trainデータの予測に失敗しました: {e}")
                    fill_value = X_train_imputed[col].median()
                    if pd.notna(fill_value):
                        X_train_imputed.loc[train_missing_idx, col] = fill_value
            
            # testデータの欠損値を予測
            if X_test_imputed is not None and col in X_test_imputed.columns and test_missing_idx.any():
                try:
                    test_missing_data = X_test_imputed.loc[test_missing_idx, feature_cols].copy()
                    test_missing_processed = test_missing_data.fillna(0)
                    predictions = self._predict(model, test_missing_processed, col)
                    X_test_imputed.loc[test_missing_idx, col] = predictions
                except Exception as e:
                    print(f"  警告: testデータの予測に失敗しました: {e}")
                    fill_value = X_train_imputed[col].median()
                    if pd.notna(fill_value):
                        X_test_imputed.loc[test_missing_idx, col] = fill_value
            
            self.imputed_columns.append(col)
            test_count = test_missing_idx.sum() if X_test_imputed is not None and hasattr(test_missing_idx, 'sum') else 0
            print(f"  完了: train {train_missing_idx.sum()}件、test {test_count}件を補完")
        
        return X_train_imputed, X_test_imputed
    
    # _preprocess_featuresメソッドは削除（数値列のみなので不要）
    
    def _train_model(self, X, y, col_name):
        """モデルの学習（数値変数のみ）"""
        if self.model_type == 'lgb':
            # 数値変数のみを補完するため、常に回帰モデルを使用
            model = lgb.LGBMRegressor(
                n_estimators=100,
                max_depth=5,
                learning_rate=0.1,
                random_state=self.random_state,
                verbose=-1,
                n_jobs=-1
            )
            
            model.fit(X, y)
            return model
        else:
            raise ValueError(f"Unknown model_type: {self.model_type}")
    
    def _predict(self, model, X, col_name):
        """予測（数値変数のみ）"""
        predictions = model.predict(X)
        # 負の値や非現実的な値はクリップ
        if col_name in ['year_built', 'floor_count']:  # 例: 年や階数は非負
            predictions = np.maximum(predictions, 0)
        
        return predictions
```

## 外れ値処理の改善

数値特徴量の外れ値を検出・処理します：
1. IQR法による外れ値検出
2. Z-score法による外れ値検出
3. 外れ値のクリッピングまたは削除


```python
# 外れ値処理の改善
print("=" * 60)
print("外れ値処理を開始します")
print("=" * 60)

# 数値列のみを対象
numeric_cols_for_outlier = X.select_dtypes(include=[np.number]).columns.tolist()

print(f"外れ値処理対象の数値列数: {len(numeric_cols_for_outlier)}")

# 外れ値処理の方法を選択（'clip' または 'remove'）
outlier_method = 'clip'  # 'clip': クリッピング, 'remove': 削除

outlier_stats = []

for col in numeric_cols_for_outlier:
    if col not in X.columns:
        continue
    
    # 欠損値が多い列はスキップ
    missing_rate = X[col].isnull().sum() / len(X)
    if missing_rate > 0.5:
        continue
    
    # 有効な値のみを使用
    valid_values = X[col].dropna()
    if len(valid_values) < 10:
        continue
    
    # IQR法による外れ値検出
    Q1 = valid_values.quantile(0.25)
    Q3 = valid_values.quantile(0.75)
    IQR = Q3 - Q1
    
    if IQR == 0:
        continue
    
    lower_bound = Q1 - 3.0 * IQR  # より緩い閾値（3.0倍）
    upper_bound = Q3 + 3.0 * IQR
    
    # 外れ値の数をカウント
    outliers_count = ((X[col] < lower_bound) | (X[col] > upper_bound)).sum()
    outlier_rate = outliers_count / len(X[col]) * 100
    
    if outliers_count > 0 and outlier_rate < 10:  # 外れ値が10%未満の場合のみ処理
        if outlier_method == 'clip':
            # クリッピング
            X.loc[X[col] < lower_bound, col] = lower_bound
            X.loc[X[col] > upper_bound, col] = upper_bound
            action = 'clipped'
        else:
            # 削除（欠損値として扱う）
            X.loc[(X[col] < lower_bound) | (X[col] > upper_bound), col] = np.nan
            action = 'removed'
        
        outlier_stats.append({
            'column': col,
            'outliers': outliers_count,
            'outlier_rate': outlier_rate,
            'lower_bound': lower_bound,
            'upper_bound': upper_bound,
            'action': action
        })

if len(outlier_stats) > 0:
    outlier_df = pd.DataFrame(outlier_stats)
    print(f"\n外れ値処理を実行した列数: {len(outlier_stats)}")
    print("\n外れ値処理の詳細（上位10列）:")
    print(outlier_df.head(10).to_string(index=False))
else:
    print("\n外れ値処理を実行した列はありませんでした")

# 目的変数の外れ値も確認（参考情報のみ、削除はしない）
if 'money_room' in df.columns:
    print("\n" + "=" * 60)
    print("目的変数（money_room）の外れ値確認")
    print("=" * 60)
    
    y_valid = y.dropna()
    Q1_y = y_valid.quantile(0.25)
    Q3_y = y_valid.quantile(0.75)
    IQR_y = Q3_y - Q1_y
    lower_bound_y = Q1_y - 3.0 * IQR_y
    upper_bound_y = Q3_y + 3.0 * IQR_y
    
    outliers_y = ((y < lower_bound_y) | (y > upper_bound_y)).sum()
    print(f"目的変数の外れ値数: {outliers_y} ({outliers_y/len(y)*100:.2f}%)")
    print(f"目的変数の範囲: [{y_valid.min():,.0f}, {y_valid.max():,.0f}]")
    print(f"IQR法による範囲: [{lower_bound_y:,.0f}, {upper_bound_y:,.0f}]")

print("\n" + "=" * 60)
print("外れ値処理が完了しました")
print("=" * 60)
```

## MAPE評価指標の実装

MAPE（Mean Absolute Percentage Error）を評価指標として使用します。
MAPE = (1/n) * Σ|actual - predicted| / |actual| * 100


```python
# MAPE評価指標の実装
print("=" * 60)
print("MAPE評価指標の実装")
print("=" * 60)

def mape_score(y_true, y_pred):
    """
    MAPE（Mean Absolute Percentage Error）を計算
    
    Parameters:
    -----------
    y_true : array-like
        実際の値
    y_pred : array-like
        予測値
    
    Returns:
    --------
    mape : float
        MAPE値（パーセント）
    """
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    
    # 0で除算を避けるため、小さい値を追加
    epsilon = 1e-8
    y_true = np.where(np.abs(y_true) < epsilon, epsilon, y_true)
    
    # MAPEの計算
    mape = np.mean(np.abs((y_true - y_pred) / y_true)) * 100
    
    return mape

def mape_lgb(y_pred, y_true):
    """
    LightGBM用のMAPEメトリック
    
    Parameters:
    -----------
    y_pred : array-like
        予測値
    y_true : lgb.Dataset
        LightGBMのデータセット
    
    Returns:
    --------
    name : str
        メトリック名
    score : float
        MAPEスコア
    is_higher_better : bool
        スコアが高い方が良いか（MAPEは低い方が良いのでFalse）
    """
    y_true = y_true.get_label()
    y_pred = y_pred
    
    # 0で除算を避けるため、小さい値を追加
    epsilon = 1e-8
    y_true = np.where(np.abs(y_true) < epsilon, epsilon, y_true)
    
    # MAPEの計算
    mape = np.mean(np.abs((y_true - y_pred) / y_true)) * 100
    
    return 'mape', mape, False

# テスト
y_test_true = np.array([100, 200, 300, 400, 500])
y_test_pred = np.array([110, 190, 310, 390, 510])
test_mape = mape_score(y_test_true, y_test_pred)
print(f"テストMAPE: {test_mape:.4f}%")
print("✓ MAPE評価指標の実装が完了しました")
```

## MAPEベースのハイパーパラメータチューニング

Optunaを使用してMAPEを最小化するハイパーパラメータを探索します。


```python
# MAPEベースのハイパーパラメータチューニング
print("=" * 60)
print("MAPEベースのハイパーパラメータチューニングを開始します")
print("=" * 60)

# Optunaのインポート
try:
    import optuna
    print("Optuna imported successfully")
except ImportError:
    print("Optuna not found. Installing...")
    import subprocess
    subprocess.check_call(["pip", "install", "optuna", "-q"])
    import optuna

# データ型の確認と修正
print("\nデータ型の確認と修正中...")
object_cols_train = X_train_encoded.select_dtypes(include=['object']).columns.tolist()
if len(object_cols_train) > 0:
    print(f"警告: object型の列が残っています。削除します: {len(object_cols_train)}列")
    X_train_encoded = X_train_encoded.drop(columns=object_cols_train)
    X_test_encoded = X_test_encoded.drop(columns=[col for col in object_cols_train if col in X_test_encoded.columns], errors='ignore')

# データ型を数値型に統一
for col in X_train_encoded.columns:
    if X_train_encoded[col].dtype == 'object':
        X_train_encoded[col] = pd.to_numeric(X_train_encoded[col], errors='coerce').fillna(0)
        if col in X_test_encoded.columns:
            X_test_encoded[col] = pd.to_numeric(X_test_encoded[col], errors='coerce').fillna(0)

# testデータの列をtrainと同じ順序に揃える
X_test_encoded = X_test_encoded.reindex(columns=X_train_encoded.columns, fill_value=0)

print(f"最終的な特徴量数: Train={X_train_encoded.shape[1]}, Test={X_test_encoded.shape[1]}")

# データの分割
from sklearn.model_selection import train_test_split
X_train_split, X_val_split, y_train_split, y_val_split = train_test_split(
    X_train_encoded, y, test_size=0.2, random_state=42, shuffle=True
)

# 目的変数の対数変換設定
use_log_transform_for_tuning = True

if use_log_transform_for_tuning:
    y_train_split_log = np.log1p(y_train_split)
    y_val_split_log = np.log1p(y_val_split)
    print("\n目的変数を対数変換して学習します（log1p）")
else:
    y_train_split_log = y_train_split
    y_val_split_log = y_val_split
    print("\n目的変数は元の値で学習します")

def objective_mape(trial):
    """Optunaの目的関数（MAPEを最小化）"""
    params = {
        'objective': 'regression',
        'metric': 'rmse',  # LightGBMの内部メトリック（MAPEはカスタムメトリックとして追加）
        'boosting_type': 'gbdt',
        'num_leaves': trial.suggest_int('num_leaves', 20, 150),
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.3, log=True),
        'feature_fraction': trial.suggest_float('feature_fraction', 0.6, 1.0),
        'bagging_fraction': trial.suggest_float('bagging_fraction', 0.6, 1.0),
        'bagging_freq': trial.suggest_int('bagging_freq', 1, 10),
        'min_child_samples': trial.suggest_int('min_child_samples', 5, 100),
        'reg_alpha': trial.suggest_float('reg_alpha', 1e-8, 10.0, log=True),
        'reg_lambda': trial.suggest_float('reg_lambda', 1e-8, 10.0, log=True),
        'verbose': -1,
        'random_state': 42
    }
    
    # データセットの作成
    train_data = lgb.Dataset(X_train_split, label=y_train_split_log)
    val_data = lgb.Dataset(X_val_split, label=y_val_split_log, reference=train_data)
    
    # モデルの学習（MAPEをカスタムメトリックとして追加）
    model = lgb.train(
        params,
        train_data,
        valid_sets=[val_data],
        valid_names=['val'],
        num_boost_round=500,
        feval=mape_lgb,
        callbacks=[lgb.early_stopping(stopping_rounds=30, verbose=False)]
    )
    
    # 検証データでの予測
    y_pred_log = model.predict(X_val_split, num_iteration=model.best_iteration)
    
    # 対数変換を使用している場合は元のスケールに戻す
    if use_log_transform_for_tuning:
        y_pred = np.expm1(y_pred_log)
        y_val_original = y_val_split
    else:
        y_pred = y_pred_log
        y_val_original = y_val_split
    
    # MAPEを計算
    mape = mape_score(y_val_original, y_pred)
    
    return mape

# Optunaで最適化（MAPEを最小化）
print("\nOptunaで最適化中（試行回数: 50、MAPEを最小化）...")
study_mape = optuna.create_study(direction='minimize')
study_mape.optimize(objective_mape, n_trials=50, show_progress_bar=True)

print(f"\n最適なパラメータ:")
print(study_mape.best_params)
print(f"最適なMAPE: {study_mape.best_value:.4f}%")

# 最適なパラメータでモデルを学習
print("\n" + "=" * 60)
print("最適なパラメータでモデルを学習中（MAPE最適化）...")
print("=" * 60)

best_params_mape = study_mape.best_params.copy()
best_params_mape.update({
    'objective': 'regression',
    'metric': 'rmse',
    'boosting_type': 'gbdt',
    'verbose': -1,
    'random_state': 42
})

train_data = lgb.Dataset(X_train_split, label=y_train_split_log)
val_data = lgb.Dataset(X_val_split, label=y_val_split_log, reference=train_data)

model_lgb_mape = lgb.train(
    best_params_mape,
    train_data,
    valid_sets=[train_data, val_data],
    valid_names=['train', 'val'],
    num_boost_round=2000,
    feval=mape_lgb,
    callbacks=[lgb.early_stopping(stopping_rounds=50), lgb.log_evaluation(period=100)]
)

print("\nMAPE最適化モデルの学習が完了しました")

# 検証データでの評価（MAPEを含む）
y_val_pred_mape_log = model_lgb_mape.predict(X_val_split, num_iteration=model_lgb_mape.best_iteration)

# 対数変換を使用している場合は元のスケールに戻す
if use_log_transform_for_tuning:
    y_val_pred_mape = np.expm1(y_val_pred_mape_log)
else:
    y_val_pred_mape = y_val_pred_mape_log

rmse_mape = np.sqrt(mean_squared_error(y_val_split, y_val_pred_mape))
mae_mape = mean_absolute_error(y_val_split, y_val_pred_mape)
r2_mape = r2_score(y_val_split, y_val_pred_mape)
mape_mape = mape_score(y_val_split, y_val_pred_mape)

print("\nMAPE最適化モデルの評価結果:")
print(f"RMSE: {rmse_mape:,.2f}")
print(f"MAE: {mae_mape:,.2f}")
print(f"R2 Score: {r2_mape:.4f}")
print(f"MAPE: {mape_mape:.4f}%")
```

## 過学習対策の実装

過学習を防ぐための対策：
1. 正則化パラメータの強化
2. より厳格なearly stopping
3. より保守的なハイパーパラメータ
4. 交差検証ベースのモデル選択
5. 学習曲線の可視化


```python
# 過学習対策の実装
print("=" * 60)
print("過学習対策を実装します")
print("=" * 60)

# 学習曲線の確認（過学習の検出）
print("\n1. 学習曲線の確認...")

# 既存のモデルの学習曲線を確認
if 'model_lgb_mape' in globals():
    print("\nMAPE最適化モデルの学習曲線:")
    print("  - train's mape: 0.0717 (約7.2%)")
    print("  - val's mape: 0.2548 (約25.5%)")
    print("  - 差: 約18.3% → 過学習の可能性が高い")
    print("\n対策が必要です。")

# 過学習対策を適用したモデルの学習
print("\n" + "=" * 60)
print("過学習対策を適用したモデルの学習")
print("=" * 60)

# データの分割（既に分割済みの場合は使用）
if 'X_train_split' not in globals() or 'X_val_split' not in globals():
    from sklearn.model_selection import train_test_split
    X_train_split, X_val_split, y_train_split, y_val_split = train_test_split(
        X_train_encoded, y, test_size=0.2, random_state=42, shuffle=True
    )

# 目的変数の対数変換
use_log_transform_anti_overfit = True

if use_log_transform_anti_overfit:
    y_train_split_log = np.log1p(y_train_split)
    y_val_split_log = np.log1p(y_val_split)
    print("\n目的変数を対数変換して学習します（log1p）")
else:
    y_train_split_log = y_train_split
    y_val_split_log = y_val_split

# 過学習対策を適用したパラメータ
# 1. 正則化の強化
# 2. より保守的なハイパーパラメータ
# 3. より厳格なearly stopping

anti_overfit_params = {
    'objective': 'regression',
    'metric': 'rmse',
    'boosting_type': 'gbdt',
    'num_leaves': 31,  # より少ない葉の数（過学習を防ぐ）
    'learning_rate': 0.01,  # より低い学習率（過学習を防ぐ）
    'feature_fraction': 0.7,  # 特徴量の70%のみ使用（過学習を防ぐ）
    'bagging_fraction': 0.7,  # データの70%のみ使用（過学習を防ぐ）
    'bagging_freq': 5,
    'min_child_samples': 50,  # より多くのサンプルが必要（過学習を防ぐ）
    'reg_alpha': 1.0,  # L1正則化を強化
    'reg_lambda': 1.0,  # L2正則化を強化
    'max_depth': 7,  # 最大深度を制限
    'min_split_gain': 0.1,  # 分割の最小ゲインを設定
    'subsample': 0.7,  # サブサンプリング率
    'colsample_bytree': 0.7,  # 特徴量サンプリング率
    'verbose': -1,
    'random_state': 42
}

print("\n過学習対策パラメータ:")
print(f"  - num_leaves: {anti_overfit_params['num_leaves']} (少なめ)")
print(f"  - learning_rate: {anti_overfit_params['learning_rate']} (低め)")
print(f"  - feature_fraction: {anti_overfit_params['feature_fraction']} (70%)")
print(f"  - bagging_fraction: {anti_overfit_params['bagging_fraction']} (70%)")
print(f"  - min_child_samples: {anti_overfit_params['min_child_samples']} (多め)")
print(f"  - reg_alpha: {anti_overfit_params['reg_alpha']} (強化)")
print(f"  - reg_lambda: {anti_overfit_params['reg_lambda']} (強化)")
print(f"  - max_depth: {anti_overfit_params['max_depth']} (制限)")

# データセットの作成
train_data_anti = lgb.Dataset(X_train_split, label=y_train_split_log)
val_data_anti = lgb.Dataset(X_val_split, label=y_val_split_log, reference=train_data_anti)

# モデルの学習（より厳格なearly stopping）
print("\nモデルを学習中（過学習対策適用）...")
model_lgb_anti_overfit = lgb.train(
    anti_overfit_params,
    train_data_anti,
    valid_sets=[train_data_anti, val_data_anti],
    valid_names=['train', 'val'],
    num_boost_round=3000,  # より多くのイテレーションを許可
    feval=mape_lgb,
    callbacks=[
        lgb.early_stopping(stopping_rounds=100, verbose=True),  # より厳格なearly stopping
        lgb.log_evaluation(period=100)
    ]
)

print("\n過学習対策モデルの学習が完了しました")

# 検証データでの評価
y_val_pred_anti_log = model_lgb_anti_overfit.predict(X_val_split, num_iteration=model_lgb_anti_overfit.best_iteration)

# 対数変換を使用している場合は元のスケールに戻す
if use_log_transform_anti_overfit:
    y_val_pred_anti = np.expm1(y_val_pred_anti_log)
else:
    y_val_pred_anti = y_val_pred_anti_log

rmse_anti = np.sqrt(mean_squared_error(y_val_split, y_val_pred_anti))
mae_anti = mean_absolute_error(y_val_split, y_val_pred_anti)
r2_anti = r2_score(y_val_split, y_val_pred_anti)
mape_anti = mape_score(y_val_split, y_val_pred_anti)

print("\n過学習対策モデルの評価結果:")
print(f"RMSE: {rmse_anti:,.2f}")
print(f"MAE: {mae_anti:,.2f}")
print(f"R2 Score: {r2_anti:.4f}")
print(f"MAPE: {mape_anti:.4f}%")

# 学習曲線の可視化
print("\n学習曲線を確認中...")
train_mape_history = []
val_mape_history = []

for i in range(100, model_lgb_anti_overfit.best_iteration + 1, 100):
    y_train_pred_i = model_lgb_anti_overfit.predict(X_train_split, num_iteration=i)
    y_val_pred_i = model_lgb_anti_overfit.predict(X_val_split, num_iteration=i)
    
    if use_log_transform_anti_overfit:
        y_train_pred_i = np.expm1(y_train_pred_i)
        y_val_pred_i = np.expm1(y_val_pred_i)
        y_train_original = y_train_split
        y_val_original = y_val_split
    else:
        y_train_original = y_train_split
        y_val_original = y_val_split
    
    train_mape_i = mape_score(y_train_original, y_train_pred_i)
    val_mape_i = mape_score(y_val_original, y_val_pred_i)
    
    train_mape_history.append(train_mape_i)
    val_mape_history.append(val_mape_i)

# 学習曲線を可視化
import matplotlib.pyplot as plt
iterations = range(100, model_lgb_anti_overfit.best_iteration + 1, 100)

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(iterations, train_mape_history, label='Train MAPE', marker='o')
plt.plot(iterations, val_mape_history, label='Val MAPE', marker='s')
plt.xlabel('Iteration')
plt.ylabel('MAPE (%)')
plt.title('Learning Curve (MAPE)')
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
gap = [v - t for t, v in zip(train_mape_history, val_mape_history)]
plt.plot(iterations, gap, label='Val - Train Gap', marker='o', color='red')
plt.xlabel('Iteration')
plt.ylabel('MAPE Gap (%)')
plt.title('Overfitting Check (Gap between Val and Train)')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

print(f"\n最終的な学習曲線:")
print(f"  Train MAPE: {train_mape_history[-1]:.4f}%")
print(f"  Val MAPE: {val_mape_history[-1]:.4f}%")
print(f"  Gap: {gap[-1]:.4f}%")

if gap[-1] > 5.0:
    print("  ⚠️ 警告: TrainとValの差が大きいです（過学習の可能性）")
else:
    print("  ✓ TrainとValの差は許容範囲内です")
```

## 目的変数の変換と交差検証

精度向上のため、以下を実施：
1. 目的変数の対数変換（分布の正規化）
2. K-Fold交差検証によるモデル評価
3. より信頼性の高い評価指標の計算


```python
# 目的変数の変換と交差検証
print("=" * 60)
print("目的変数の変換と交差検証を開始します")
print("=" * 60)

# 目的変数の分布を確認
print("\n目的変数（money_room）の分布確認:")
print(f"  平均: {y.mean():,.2f}")
print(f"  中央値: {y.median():,.2f}")
print(f"  標準偏差: {y.std():,.2f}")
print(f"  最小値: {y.min():,.2f}")
print(f"  最大値: {y.max():,.2f}")
print(f"  歪度: {y.skew():.4f}")

# 目的変数の対数変換（分布が右に歪んでいる場合）
use_log_transform = True  # True: 対数変換を使用, False: 元の値を使用

if use_log_transform:
    # 対数変換（log1pを使用して0を扱えるようにする）
    y_original = y.copy()
    y = np.log1p(y)
    print("\n目的変数を対数変換しました（log1p）")
    print(f"  変換後の平均: {y.mean():,.2f}")
    print(f"  変換後の標準偏差: {y.std():,.2f}")
    print(f"  変換後の歪度: {y.skew():.4f}")
else:
    print("\n目的変数は変換しません（元の値を使用）")

# K-Fold交差検証
print("\n" + "=" * 60)
print("K-Fold交差検証を開始します")
print("=" * 60)

from sklearn.model_selection import KFold

# 交差検証の設定
n_splits = 5  # 5-Fold CV
kf = KFold(n_splits=n_splits, shuffle=True, random_state=42)

# 交差検証用のパラメータ（簡易版、高速化のため）
cv_params = {
    'objective': 'regression',
    'metric': 'rmse',
    'boosting_type': 'gbdt',
    'num_leaves': 31,
    'learning_rate': 0.05,
    'feature_fraction': 0.9,
    'bagging_fraction': 0.8,
    'bagging_freq': 5,
    'verbose': -1,
    'random_state': 42
}

cv_scores = []
cv_models = []

print(f"\n{n_splits}-Fold交差検証を実行中...")

for fold, (train_idx, val_idx) in enumerate(kf.split(X_train_encoded), 1):
    print(f"\nFold {fold}/{n_splits} を実行中...")
    
    # データの分割
    X_train_fold = X_train_encoded.iloc[train_idx]
    X_val_fold = X_train_encoded.iloc[val_idx]
    y_train_fold = y.iloc[train_idx]
    y_val_fold = y.iloc[val_idx]
    
    # データセットの作成
    train_data_fold = lgb.Dataset(X_train_fold, label=y_train_fold)
    val_data_fold = lgb.Dataset(X_val_fold, label=y_val_fold, reference=train_data_fold)
    
    # モデルの学習
    model_cv = lgb.train(
        cv_params,
        train_data_fold,
        valid_sets=[val_data_fold],
        valid_names=['val'],
        num_boost_round=500,
        callbacks=[lgb.early_stopping(stopping_rounds=30, verbose=False)]
    )
    
    # 予測
    y_pred_fold = model_cv.predict(X_val_fold, num_iteration=model_cv.best_iteration)
    
    # 対数変換を使用している場合は元のスケールに戻す
    if use_log_transform:
        y_pred_fold = np.expm1(y_pred_fold)
        y_val_fold_original = np.expm1(y_val_fold)
    else:
        y_val_fold_original = y_val_fold
    
    # 評価指標の計算
    rmse_fold = np.sqrt(mean_squared_error(y_val_fold_original, y_pred_fold))
    mae_fold = mean_absolute_error(y_val_fold_original, y_pred_fold)
    r2_fold = r2_score(y_val_fold_original, y_pred_fold)
    
    cv_scores.append({
        'fold': fold,
        'rmse': rmse_fold,
        'mae': mae_fold,
        'r2': r2_fold
    })
    
    cv_models.append(model_cv)
    
    print(f"  Fold {fold} - RMSE: {rmse_fold:,.2f}, MAE: {mae_fold:,.2f}, R2: {r2_fold:.4f}")

# 交差検証結果の集計
cv_results_df = pd.DataFrame(cv_scores)
print("\n" + "=" * 60)
print("交差検証結果の集計")
print("=" * 60)
print(cv_results_df.to_string(index=False))

print(f"\n平均 RMSE: {cv_results_df['rmse'].mean():,.2f} ± {cv_results_df['rmse'].std():,.2f}")
print(f"平均 MAE: {cv_results_df['mae'].mean():,.2f} ± {cv_results_df['mae'].std():,.2f}")
print(f"平均 R2: {cv_results_df['r2'].mean():.4f} ± {cv_results_df['r2'].std():.4f}")

# 目的変数を元に戻す（後続の処理で使用するため）
if use_log_transform:
    y = y_original.copy()
    print("\n目的変数を元のスケールに戻しました")
```

```python
# 特徴量エンジニアリング（エンコーディング前）
print("=" * 60)
print("特徴量エンジニアリングを開始します（エンコーディング前）")
print("=" * 60)

# エンコーディング前のデータを使用
X_train_fe = X_imputed.copy()
X_test_fe = X_test_imputed.copy()

print(f"特徴量エンジニアリング前の特徴量数: Train={X_train_fe.shape[1]}, Test={X_test_fe.shape[1]}")

# 1. 日付特徴量の抽出
print("\n1. 日付特徴量の抽出...")

# year_builtから築年数を計算
if 'year_built' in X_train_fe.columns and 'target_ym' in X_train_fe.columns:
    def extract_year_month(value):
        """年月を抽出（例: 202001 -> 2020年1月）"""
        if pd.isna(value) or value == 0:
            return None, None
        try:
            year_month = int(value)
            year = year_month // 100
            month = year_month % 100
            return year, month
        except:
            return None, None
    
    # target_ymから年と月を抽出
    train_years = (X_train_fe['target_ym'] // 100).astype(int)
    train_months = (X_train_fe['target_ym'] % 100).astype(int)
    test_years = (X_test_fe['target_ym'] // 100).astype(int)
    test_months = (X_test_fe['target_ym'] % 100).astype(int)
    
    # year_builtから年と月を抽出
    train_built_info = X_train_fe['year_built'].apply(extract_year_month)
    test_built_info = X_test_fe['year_built'].apply(extract_year_month)
    
    train_built_years = train_built_info.apply(lambda x: x[0] if x[0] else None)
    train_built_months = train_built_info.apply(lambda x: x[1] if x[1] else None)
    test_built_years = test_built_info.apply(lambda x: x[0] if x[0] else None)
    test_built_months = test_built_info.apply(lambda x: x[1] if x[1] else None)
    
    # 築年数を計算（年単位）
    X_train_fe['building_age'] = train_years - train_built_years
    X_test_fe['building_age'] = test_years - test_built_years
    
    # 築年数を月単位で計算（より正確）
    X_train_fe['building_age_months'] = (train_years - train_built_years) * 12 + (train_months - train_built_months)
    X_test_fe['building_age_months'] = (test_years - test_built_years) * 12 + (test_months - test_built_months)
    
    # 負の値を0にクリップ
    X_train_fe['building_age'] = X_train_fe['building_age'].clip(lower=0)
    X_test_fe['building_age'] = X_test_fe['building_age'].clip(lower=0)
    X_train_fe['building_age_months'] = X_train_fe['building_age_months'].clip(lower=0)
    X_test_fe['building_age_months'] = X_test_fe['building_age_months'].clip(lower=0)
    
    # 欠損値を0で埋める
    X_train_fe['building_age'] = X_train_fe['building_age'].fillna(0)
    X_test_fe['building_age'] = X_test_fe['building_age'].fillna(0)
    X_train_fe['building_age_months'] = X_train_fe['building_age_months'].fillna(0)
    X_test_fe['building_age_months'] = X_test_fe['building_age_months'].fillna(0)
    
    print(f"  ✓ 築年数特徴量を作成: building_age, building_age_months")

# 2. 地理的特徴量の活用
print("\n2. 地理的特徴量の活用...")

if 'lon' in X_train_fe.columns and 'lat' in X_train_fe.columns:
    # 東京の中心座標（皇居周辺）
    tokyo_center_lon = 139.6917
    tokyo_center_lat = 35.6895
    
    # 中心からの距離を計算（簡易版：ユークリッド距離）
    X_train_fe['distance_from_center'] = np.sqrt(
        (X_train_fe['lon'] - tokyo_center_lon) ** 2 + 
        (X_train_fe['lat'] - tokyo_center_lat) ** 2
    )
    X_test_fe['distance_from_center'] = np.sqrt(
        (X_test_fe['lon'] - tokyo_center_lon) ** 2 + 
        (X_test_fe['lat'] - tokyo_center_lat) ** 2
    )
    
    # 欠損値を中央値で埋める
    if X_train_fe['distance_from_center'].notna().sum() > 0:
        median_dist = X_train_fe['distance_from_center'].median()
        X_train_fe['distance_from_center'] = X_train_fe['distance_from_center'].fillna(median_dist)
        X_test_fe['distance_from_center'] = X_test_fe['distance_from_center'].fillna(median_dist)
    
    print(f"  ✓ 中心からの距離特徴量を作成: distance_from_center")

# 3. 面積比率特徴量の作成
print("\n3. 面積比率特徴量の作成...")

# 専有面積と建物面積の比率
if 'unit_area' in X_train_fe.columns and 'building_area' in X_train_fe.columns:
    X_train_fe['unit_area_ratio'] = X_train_fe['unit_area'] / (X_train_fe['building_area'] + 1e-6)
    X_test_fe['unit_area_ratio'] = X_test_fe['unit_area'] / (X_test_fe['building_area'] + 1e-6)
    X_train_fe['unit_area_ratio'] = X_train_fe['unit_area_ratio'].fillna(0)
    X_test_fe['unit_area_ratio'] = X_test_fe['unit_area_ratio'].fillna(0)
    print(f"  ✓ 専有面積比率を作成: unit_area_ratio")

# 建物面積と土地面積の比率
if 'building_area' in X_train_fe.columns and 'land_area_all' in X_train_fe.columns:
    X_train_fe['building_land_ratio'] = X_train_fe['building_area'] / (X_train_fe['land_area_all'] + 1e-6)
    X_test_fe['building_land_ratio'] = X_test_fe['building_area'] / (X_test_fe['land_area_all'] + 1e-6)
    X_train_fe['building_land_ratio'] = X_train_fe['building_land_ratio'].fillna(0)
    X_test_fe['building_land_ratio'] = X_test_fe['building_land_ratio'].fillna(0)
    print(f"  ✓ 建物土地比率を作成: building_land_ratio")

# 専有面積と総戸数の比率（1戸あたりの面積）
if 'unit_area' in X_train_fe.columns and 'unit_count' in X_train_fe.columns:
    X_train_fe['unit_area_per_unit'] = X_train_fe['unit_area'] / (X_train_fe['unit_count'] + 1e-6)
    X_test_fe['unit_area_per_unit'] = X_test_fe['unit_area'] / (X_test_fe['unit_count'] + 1e-6)
    X_train_fe['unit_area_per_unit'] = X_train_fe['unit_area_per_unit'].fillna(0)
    X_test_fe['unit_area_per_unit'] = X_test_fe['unit_area_per_unit'].fillna(0)
    print(f"  ✓ 1戸あたりの面積を作成: unit_area_per_unit")

# 4. タグ情報の処理
print("\n4. タグ情報の処理...")

for tag_col in ['building_tag_id', 'unit_tag_id']:
    if tag_col in X_train_fe.columns:
        # タグの数をカウント（スラッシュ区切り）
        X_train_fe[f'{tag_col}_count'] = X_train_fe[tag_col].astype(str).apply(
            lambda x: len(str(x).split('/')) if pd.notna(x) and str(x) != 'nan' and '/' in str(x) else 0
        )
        X_test_fe[f'{tag_col}_count'] = X_test_fe[tag_col].astype(str).apply(
            lambda x: len(str(x).split('/')) if pd.notna(x) and str(x) != 'nan' and '/' in str(x) else 0
        )
        print(f"  ✓ {tag_col}のタグ数をカウント: {tag_col}_count")

# 5. 数値特徴量の変換
print("\n5. 数値特徴量の変換...")

# 対数変換（正の値のみ）
numeric_cols_for_log = ['unit_area', 'building_area', 'total_floor_area', 'land_area_all', 'money_shuuzen']
for col in numeric_cols_for_log:
    if col in X_train_fe.columns:
        # 正の値のみ対数変換
        X_train_fe[f'{col}_log'] = np.log1p(X_train_fe[col].clip(lower=0))
        X_test_fe[f'{col}_log'] = np.log1p(X_test_fe[col].clip(lower=0))
        X_train_fe[f'{col}_log'] = X_train_fe[f'{col}_log'].fillna(0)
        X_test_fe[f'{col}_log'] = X_test_fe[f'{col}_log'].fillna(0)
        print(f"  ✓ {col}の対数変換: {col}_log")

# 2乗項（重要な特徴量のみ）
important_cols = ['unit_area', 'building_area', 'year_built', 'floor_count', 'building_age']
for col in important_cols:
    if col in X_train_fe.columns:
        X_train_fe[f'{col}_squared'] = X_train_fe[col] ** 2
        X_test_fe[f'{col}_squared'] = X_test_fe[col] ** 2
        X_train_fe[f'{col}_squared'] = X_train_fe[f'{col}_squared'].fillna(0)
        X_test_fe[f'{col}_squared'] = X_test_fe[f'{col}_squared'].fillna(0)
        print(f"  ✓ {col}の2乗項: {col}_squared")

print("\n" + "=" * 60)
print("特徴量エンジニアリングが完了しました")
print("=" * 60)
print(f"特徴量数: Train={X_train_fe.shape[1]}, Test={X_test_fe.shape[1]}")

# 更新
X_imputed = X_train_fe.copy()
X_test_imputed = X_test_fe.copy()
```

```python
# 特徴量エンジニアリングの改善
print("=" * 60)
print("特徴量エンジニアリングを開始します")
print("=" * 60)

# エンコーディング前のデータを使用
X_train_fe = X_imputed.copy()
X_test_fe = X_test_imputed.copy()

# 1. 日付特徴量の抽出
print("\n1. 日付特徴量の抽出...")

# year_builtから年齢を計算
if 'year_built' in X_train_fe.columns:
    # year_builtを年月に変換（例: 202001 -> 2020年1月）
    def extract_year_month(value):
        if pd.isna(value) or value == 0:
            return None, None
        try:
            year_month = int(value)
            year = year_month // 100
            month = year_month % 100
            return year, month
        except:
            return None, None
    
    # target_ymから年と月を抽出
    if 'target_ym' in X_train_fe.columns:
        train_years = (X_train_fe['target_ym'] // 100).astype(int)
        train_months = (X_train_fe['target_ym'] % 100).astype(int)
        test_years = (X_test_fe['target_ym'] // 100).astype(int)
        test_months = (X_test_fe['target_ym'] % 100).astype(int)
        
        # year_builtから年と月を抽出
        train_built_years = X_train_fe['year_built'].apply(lambda x: extract_year_month(x)[0] if extract_year_month(x)[0] else None)
        train_built_months = X_train_fe['year_built'].apply(lambda x: extract_year_month(x)[1] if extract_year_month(x)[1] else None)
        test_built_years = X_test_fe['year_built'].apply(lambda x: extract_year_month(x)[0] if extract_year_month(x)[0] else None)
        test_built_months = X_test_fe['year_built'].apply(lambda x: extract_year_month(x)[1] if extract_year_month(x)[1] else None)
        
        # 築年数を計算（年単位）
        X_train_fe['building_age'] = train_years - train_built_years
        X_test_fe['building_age'] = test_years - test_built_years
        
        # 築年数を月単位で計算（より正確）
        X_train_fe['building_age_months'] = (train_years - train_built_years) * 12 + (train_months - train_built_months)
        X_test_fe['building_age_months'] = (test_years - test_built_years) * 12 + (test_months - test_built_months)
        
        # 負の値を0にクリップ
        X_train_fe['building_age'] = X_train_fe['building_age'].clip(lower=0)
        X_test_fe['building_age'] = X_test_fe['building_age'].clip(lower=0)
        X_train_fe['building_age_months'] = X_train_fe['building_age_months'].clip(lower=0)
        X_test_fe['building_age_months'] = X_test_fe['building_age_months'].clip(lower=0)
        
        print(f"  ✓ 築年数特徴量を作成: building_age, building_age_months")

# 2. 地理的特徴量の活用
print("\n2. 地理的特徴量の活用...")

if 'lon' in X_train_fe.columns and 'lat' in X_train_fe.columns:
    # 東京の中心座標（例: 皇居周辺）
    tokyo_center_lon = 139.6917
    tokyo_center_lat = 35.6895
    
    # 中心からの距離を計算（簡易版：ユークリッド距離）
    X_train_fe['distance_from_center'] = np.sqrt(
        (X_train_fe['lon'] - tokyo_center_lon) ** 2 + 
        (X_train_fe['lat'] - tokyo_center_lat) ** 2
    )
    X_test_fe['distance_from_center'] = np.sqrt(
        (X_test_fe['lon'] - tokyo_center_lon) ** 2 + 
        (X_test_fe['lat'] - tokyo_center_lat) ** 2
    )
    
    print(f"  ✓ 中心からの距離特徴量を作成: distance_from_center")

# 3. 面積比率特徴量の作成
print("\n3. 面積比率特徴量の作成...")

# 専有面積と建物面積の比率
if 'unit_area' in X_train_fe.columns and 'building_area' in X_train_fe.columns:
    X_train_fe['unit_area_ratio'] = X_train_fe['unit_area'] / (X_train_fe['building_area'] + 1e-6)
    X_test_fe['unit_area_ratio'] = X_test_fe['unit_area'] / (X_test_fe['building_area'] + 1e-6)
    print(f"  ✓ 専有面積比率を作成: unit_area_ratio")

# 建物面積と土地面積の比率
if 'building_area' in X_train_fe.columns and 'land_area_all' in X_train_fe.columns:
    X_train_fe['building_land_ratio'] = X_train_fe['building_area'] / (X_train_fe['land_area_all'] + 1e-6)
    X_test_fe['building_land_ratio'] = X_test_fe['building_area'] / (X_test_fe['land_area_all'] + 1e-6)
    print(f"  ✓ 建物土地比率を作成: building_land_ratio")

# 4. タグ情報の処理
print("\n4. タグ情報の処理...")

for tag_col in ['building_tag_id', 'unit_tag_id']:
    if tag_col in X_train_fe.columns:
        # タグの数をカウント（スラッシュ区切り）
        X_train_fe[f'{tag_col}_count'] = X_train_fe[tag_col].astype(str).apply(
            lambda x: len(str(x).split('/')) if pd.notna(x) and x != 'nan' and '/' in str(x) else 0
        )
        X_test_fe[f'{tag_col}_count'] = X_test_fe[tag_col].astype(str).apply(
            lambda x: len(str(x).split('/')) if pd.notna(x) and x != 'nan' and '/' in str(x) else 0
        )
        print(f"  ✓ {tag_col}のタグ数をカウント: {tag_col}_count")

# 5. 数値特徴量の変換
print("\n5. 数値特徴量の変換...")

# 対数変換（正の値のみ）
numeric_cols_for_log = ['unit_area', 'building_area', 'total_floor_area', 'land_area_all']
for col in numeric_cols_for_log:
    if col in X_train_fe.columns:
        # 正の値のみ対数変換
        X_train_fe[f'{col}_log'] = np.log1p(X_train_fe[col].clip(lower=0))
        X_test_fe[f'{col}_log'] = np.log1p(X_test_fe[col].clip(lower=0))
        print(f"  ✓ {col}の対数変換: {col}_log")

# 2乗項（重要な特徴量のみ）
important_cols = ['unit_area', 'building_area', 'year_built', 'floor_count']
for col in important_cols:
    if col in X_train_fe.columns:
        X_train_fe[f'{col}_squared'] = X_train_fe[col] ** 2
        X_test_fe[f'{col}_squared'] = X_test_fe[col] ** 2
        print(f"  ✓ {col}の2乗項: {col}_squared")

print("\n" + "=" * 60)
print("特徴量エンジニアリングが完了しました")
print("=" * 60)
print(f"特徴量数: Train={X_train_fe.shape[1]}, Test={X_test_fe.shape[1]}")

# 更新
X_imputed = X_train_fe.copy()
X_test_imputed = X_test_fe.copy()
```

```python
# データの分割（検証用）
from sklearn.model_selection import train_test_split, cross_val_score, KFold
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# train/validationに分割
X_train_split, X_val_split, y_train_split, y_val_split = train_test_split(
    X_train_encoded, y, test_size=0.2, random_state=42, shuffle=True
)

print(f"Train分割後の形状: {X_train_split.shape}")
print(f"Validation分割後の形状: {X_val_split.shape}")
```

```python
# LightGBMモデルの構築
print("LightGBMモデルの学習を開始します...")

# モデルのパラメータ
lgb_params = {
    'objective': 'regression',
    'metric': 'rmse',
    'boosting_type': 'gbdt',
    'num_leaves': 31,
    'learning_rate': 0.05,
    'feature_fraction': 0.9,
    'bagging_fraction': 0.8,
    'bagging_freq': 5,
    'verbose': -1,
    'random_state': 42
}

# データセットの作成
train_data = lgb.Dataset(X_train_split, label=y_train_split)
val_data = lgb.Dataset(X_val_split, label=y_val_split, reference=train_data)

# モデルの学習
model_lgb = lgb.train(
    lgb_params,
    train_data,
    valid_sets=[train_data, val_data],
    valid_names=['train', 'val'],
    num_boost_round=1000,
    callbacks=[lgb.early_stopping(stopping_rounds=50), lgb.log_evaluation(period=100)]
)

print("\nモデルの学習が完了しました")
```

```python
# 検証データでの予測と評価
y_val_pred = model_lgb.predict(X_val_split, num_iteration=model_lgb.best_iteration)

# 評価指標の計算
rmse = np.sqrt(mean_squared_error(y_val_split, y_val_pred))
mae = mean_absolute_error(y_val_split, y_val_pred)
r2 = r2_score(y_val_split, y_val_pred)

print("検証データでの評価結果:")
print(f"RMSE: {rmse:,.2f}")
print(f"MAE: {mae:,.2f}")
print(f"R2 Score: {r2:.4f}")

# 予測値と実際の値の可視化
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.scatter(y_val_split, y_val_pred, alpha=0.3)
plt.plot([y_val_split.min(), y_val_split.max()], [y_val_split.min(), y_val_split.max()], 'r--', lw=2)
plt.xlabel('Actual money_room')
plt.ylabel('Predicted money_room')
plt.title(f'Prediction vs Actual (R2={r2:.4f})')

plt.subplot(1, 2, 2)
residuals = y_val_split - y_val_pred
plt.scatter(y_val_pred, residuals, alpha=0.3)
plt.axhline(y=0, color='r', linestyle='--')
plt.xlabel('Predicted money_room')
plt.ylabel('Residuals')
plt.title('Residual Plot')

plt.tight_layout()
plt.show()
```

```python
import os

# sample_submit.csvを参考に提出ファイルを作成

print("=" * 60)
print("提出ファイルの作成を開始します")
print("=" * 60)

# sample_submit.csvのカラム名を取得（id, money_room）し、テスト予測に適用する
sample_submit_path = '/workspaces/competition/data/input/20251114/sample_submit.csv'
sample_submit = pd.read_csv(sample_submit_path, nrows=5)
submit_cols = sample_submit.columns.tolist()

print(f"\n予測値の数: {len(y_test_pred)}")
print(f"testデータの行数: {len(df_test)}")

# テストデータのID列取得（サンプルと同じ順番で出力するためにdf_test['id']参照）
if 'id' in df_test.columns:
    test_ids = df_test['id'].values.copy()
    print(f"testデータのIDの数: {len(test_ids)}")
    print(f"testデータのIDの型: {type(test_ids[0])}")
    print(f"testデータのIDの最初の5個: {test_ids[:5]}")
else:
    # 万が一、カラム名が違う/存在しない場合に備えて
    test_ids = np.arange(len(y_test_pred))
    print("警告: 'id'列が見つかりません。連番を使用します。")

# 予測値の数とtestデータのIDの数を確認
if len(y_test_pred) != len(test_ids):
    print(f"\n警告: 予測値の数({len(y_test_pred)})とtestデータのIDの数({len(test_ids)})が一致しません。")
    # 少ない方に合わせる
    min_len = min(len(y_test_pred), len(test_ids))
    y_test_pred = y_test_pred[:min_len]
    test_ids = test_ids[:min_len]
    print(f"調整後の数: {min_len}")
else:
    print(f"\n✓ 予測値の数とtestデータのIDの数が一致しています: {len(y_test_pred)}")

# 6桁ゼロ詰めのIDを作る（文字列として扱う）
id_strs = []
for i, x in enumerate(test_ids):
    try:
        # 数値の場合はintに変換してから6桁ゼロ詰め
        if pd.isna(x):
            id_strs.append("000000")
        else:
            # 数値型の場合
            if isinstance(x, (int, float, np.integer, np.floating)):
                id_strs.append(f"{int(float(x)):06d}")
            else:
                # 文字列の場合
                id_str = str(x).strip()
                if id_str.isdigit():
                    id_strs.append(f"{int(id_str):06d}")
                elif len(id_str) == 6 and id_str.isdigit():
                    id_strs.append(id_str)
                else:
                    # フォールバック: インデックスを使用
                    id_strs.append(f"{i:06d}")
    except (ValueError, TypeError) as e:
        # エラーが発生した場合はインデックスを使用
        id_strs.append(f"{i:06d}")

# 予測値は四捨五入してintに変換
money_values = np.round(y_test_pred).astype(int)

# サンプルと同じカラム順でDataFrame作成
submission = pd.DataFrame({
    submit_cols[0]: id_strs,
    submit_cols[1]: money_values
})

# IDの重複を確認
print(f"\n提出ファイルの行数: {len(submission)}")
print(f"提出ファイルのIDのユニーク数: {submission[submit_cols[0]].nunique()}")
duplicated_ids = submission[submit_cols[0]].value_counts()
if (duplicated_ids > 1).any():
    print(f"\n警告: IDの重複があります:")
    print(duplicated_ids[duplicated_ids > 1].head(10))
    print("\n重複を修正します...")
    # 重複を修正（連番で置き換え）
    unique_ids = []
    seen = set()
    for id_str in id_strs:
        if id_str in seen:
            # 重複している場合は新しいIDを生成
            new_id = f"{len(unique_ids):06d}"
            while new_id in seen:
                new_id = f"{len(unique_ids):06d}"
            unique_ids.append(new_id)
            seen.add(new_id)
        else:
            unique_ids.append(id_str)
            seen.add(id_str)
    
    submission[submit_cols[0]] = unique_ids
    print("重複を修正しました")
else:
    print("✓ IDの重複はありません")

# サンプル出力確認
print("\n提出ファイル例（最初の10行）:")
print(submission.head(10))

# 提出ファイル保存パス生成
output_dir = '/workspaces/competition/data/output/20251114'
os.makedirs(output_dir, exist_ok=True)
output_path = os.path.join(output_dir, f'submission_{timestamp}.csv')

# CSV保存（ヘッダーなし、sample_submit.csvと同じ形式）
submission.to_csv(output_path, index=False, header=False)
print(f'\n提出ファイルを保存しました: {output_path}')

# 保存したファイルを確認
print("\n保存したファイルの最初の5行を確認:")
with open(output_path, 'r') as f:
    for i, line in enumerate(f):
        if i < 5:
            print(line.strip())
        else:
            break

print("\n" + "=" * 60)
print("提出ファイルの作成が完了しました")
print("=" * 60)
```
